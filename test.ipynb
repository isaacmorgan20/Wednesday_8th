```python
import pandas as pd
import numpy as np
```

```python
data = {
    "employee_id": [101, 102, 103, 104, 105, 102, 106, 107, 108, 109],
    "name": ["Alice Johnson", "bob smith", "Charlie Brown", "Diana", "Eve Adams", "bob smith", "Frank ", " Grace Lee", "Hank", "Ivy Clarke"],
    "department": ["HR", "IT", "Finance", "hr", "IT", "IT", "FINANCE", "HR", None, "Finance"],
    "age": [29, 34, 45, 27, None, 34, 300, 38, 25, 31],
    "salary": ["$52,000", "$78,000", "$91,000", "$48,500", "$85,000", "$78,000", "$102,000", "$67,000", "$54,000", None],
    "join_date": ["2021-03-15", "2019-07-22", "2018-01-10", "2022-11-05", "2020-06-30", "2019-07-22", "2017-09-01", "2023-02-14", "2021-08-19", "2022-04-03"],
    "rating": [4.2, 3.8, 4.5, None, 4.0, 3.8, 4.1, 3.5, None, 4.3]
    }
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
# displays the first five rows
# df.head()

# displays the last five rows
# df.tail()

# displays the number of rows and columns in a dataframe (row,columns)
# df.shape

# df.info()

# df.describe()

df.columns
# df.dtypes
```

```python
# To check for date
df["join_date"]
```

```python
# To check for duplicates
print(df[df.duplicated(keep=False)])
```

```python
# To remove space and extra space in the name column
df["name"] = df["name"].str.strip().str.replace(r"/s+", " ", regex=True)

# To change the first letters of the names from lower case to upper case in the name column
df["name"] = df["name"].str.title()

# To remove duplicates from the data
df = df.drop_duplicates()

# To check for name
print(df["name"].tolist())
```

```python
# To change the first letters of department from lower case to upper case in the department column
df["department"] = df["department"].str.upper()

# Fill the missing column of department with UNKNOWN in the department column
df["department"] = df["department"].fillna("UNKNOWN")

# Check for department in list form
df["department"].tolist()
```

```python
# A program to change age greater than 100 to a NaN
df.loc[df["age"] > 100, "age"] = np.nan

# Finding the median in the age and storing it in a variable median_age
median_age = df["age"].median()

# Filling the missing value with the median_age
df["age"] = df["age"].fillna(median_age)

# Print the median_age
print(median_age)
```

```python
# Replacing the "$" with nothing and "," with nothing in the salary column
df["salary"] = df["salary"].str.replace("$", "").str.replace(",","")

# Changing the salary from string to float
df["salary"] = df["salary"].astype(float)

# Finding the median in the salary and storing it in a variable median_salary
median_salary = df["salary"].median()
 
# Print the median_salary
print(median_salary)

# Filling the missing value with the median_salary
df["salary"] = df["salary"].fillna(median_salary)

# Print salary
print(df["salary"])
```

```python
# Converting date from string to real date
df["join_date"] = pd.to_datetime(df["join_date"])
```

```python
# Check for the data type of the date
print(df["join_date"].dtypes)
```

```python
# Finding the median in the rating and storing it in a variable median_rating
median_rating = df["rating"].median()

# Print median_rating
print(median_rating)

# Filling the missing value with the median_rating
df["rating"] = df["rating"].fillna(median_rating)

# Print rating
print(df["rating"])
```

```python
# Check the data frame
print(df)
```

```python
# Check for the data type of the data
print(df["join_date"].dtypes)
```

```python
# Save the data in csv format setting index to false (excluding the index)
df.to_csv("clean.csv", index=False)
```
